# Plot Denemeleri

```python
from io import StringIO
from pathlib import Path
import pandas as pd
import numpy as np
from datetime import timedelta
from zoneinfo import ZoneInfo
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

CATALOG_DATA = """
10/11/2025 18:59:01,1.9,Sındırgı (Balıkesir)
10/11/2025 18:52:13,1.3,Sındırgı (Balıkesir)
10/11/2025 18:46:35,1.3,Sındırgı (Balıkesir)
10/11/2025 18:38:17,2.3,Sındırgı (Balıkesir)
10/11/2025 18:35:50,1.8,Sındırgı (Balıkesir)
10/11/2025 18:30:55,1.4,Sındırgı (Balıkesir)
10/11/2025 18:27:03,1.9,Sındırgı (Balıkesir)
10/11/2025 18:20:09,3.9,Sındırgı (Balıkesir)
""".strip()

csv_path = Path("/home/ege/recovar/reproducibility/phasenet_eqt/10NOVEMBER-18-19/10NOVEMBER_INSTANCE/KO_BALB_predictions.csv")
station_name = "BALB"
TIMEZONE = "Europe/Istanbul"
tz = ZoneInfo(TIMEZONE)
PLOT_START = pd.to_datetime("2025-11-10 18:10:00").tz_localize(tz)
PLOT_END   = pd.to_datetime("2025-11-10 19:00:00").tz_localize(tz)

df = pd.read_csv(csv_path, parse_dates=["timestamp"])
predictions = pd.Series(data=df['eq_probability'].values, index=df['timestamp'])
predictions = predictions.sort_index()

catalog_df = pd.read_csv(StringIO(CATALOG_DATA), header=None, names=['datetime','magnitude','location'])
catalog_df['datetime'] = pd.to_datetime(catalog_df['datetime'], format='%d/%m/%Y %H:%M:%S')
catalog_df['datetime'] = catalog_df['datetime'].dt.tz_localize(tz)
catalog_times = catalog_df['datetime'].tolist()

plt.figure(figsize=(14, 5))
ax = plt.gca()

time_formatter = mdates.DateFormatter('%H:%M', tz=tz)
ax.xaxis.set_major_formatter(time_formatter)
ax.xaxis.set_major_locator(mdates.MinuteLocator(interval=5))
plt.setp(ax.xaxis.get_majorticklabels(), rotation=30, ha='right')

ax.fill_between(predictions.index, 0, predictions.values, alpha=0.3, color='steelblue', label=f'{station_name} (prob)')
ax.plot(predictions.index, predictions.values, color='steelblue', linewidth=1.5)

# CATALOG LINES
for idx, eq_time in enumerate(catalog_times):
    if not (PLOT_START <= eq_time <= PLOT_END):
        continue
    ax.axvline(eq_time, color='red', linestyle='--', alpha=0.6, linewidth=1.2,
               label='Catalog EQ' if idx == 0 else None)

ax.set_xlim(PLOT_START, PLOT_END)
ax.set_ylim(0, 1)
ax.set_ylabel('Event Score')
ax.set_xlabel('Time')
ax.set_title(f'Recovar Event Scores - {station_name} Date: {PLOT_START} - {PLOT_END}')
ax.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
